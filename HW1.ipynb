# Sigmoid Function Gradient

$$\sigma(x) = \frac{1}{1+e^{-x}} = (1+e^{-x})^{-1}$$

$$(f(g(x)))' = f'(g(x)) \cdot g'(x)$$

$$f(g(x)) = (1+e^{-x})^{-1} \quad \rightarrow \quad f'(g(x)) = -(1+e^{-x})^{-2}$$

$$g(x) = 1+e^{-x} \quad \rightarrow \quad g'(x) = -e^{-x}$$

$$\sigma' = f'(g(x)) \cdot g'(x) = -(1+e^{-x})^{-2} \cdot (-e^{-x}) = e^{-x}(1+e^{-x})^{-2}$$

  $\sigma' = \sigma(1-\sigma)$

$$\sigma = \frac{1}{1+e^{-x}} \quad \Rightarrow \quad \sigma(1+e^{-x}) = 1$$

$$\sigma' = e^{-x}(1+e^{-x})^{-2} = e^{-x}\left(\frac{1}{(1+e^{-x})^2}\right) = e^{-x}\left(\frac{1}{1+e^{-x}} \cdot \frac{1}{1+e^{-x}}\right)$$

$$= e^{-x} \cdot \sigma^2 = \sigma(\sigma \cdot e^{-x}) = \sigma(\sigma(e^{-x}+1-1))$$

$$= \sigma(\underbrace{\sigma(e^{-x}+1)}_{1} - \sigma) = \sigma(1-\sigma)$$

# Tanh Function Gradient

the hyperbolic tangent function:

$$\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$


$$\left(\frac{f}{g}\right)' = \frac{f' \cdot g - f \cdot g'}{g^2}$$

$$f = e^x - e^{-x}$$
$$f' = e^x + e^{-x}$$
$$g = e^x + e^{-x}$$
$$g' = e^x - e^{-x}$$

$$(\tanh)' = \frac{(e^x+e^{-x})(e^x+e^{-x}) - (e^x-e^{-x})(e^x-e^{-x})}{(e^x+e^{-x})^2} = \frac{(e^x+e^{-x})^2 - (e^x-e^{-x})^2}{(e^x+e^{-x})^2}$$

$$= \frac{(e^x+e^{-x})^2}{(e^x+e^{-x})^2} - \frac{(e^x-e^{-x})^2}{(e^x+e^{-x})^2} = 1 - \frac{(e^x-e^{-x})^2}{(e^x+e^{-x})^2}$$

$$= 1 - \left(\frac{e^x-e^{-x}}{e^x+e^{-x}}\right)^2 = 1 - \tanh^2(x)$$
